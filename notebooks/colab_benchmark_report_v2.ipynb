# OneVoice V2 — benchmark report
Đọc các `aggregate.json` đã có trên Google Drive và tạo báo cáo HTML/Markdown có biểu đồ. Không chạy lại benchmark và không thay đổi model/data.


```python
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
import os, subprocess, sys
GITHUB_REPO = 'https://github.com/Platypus27-coder/OneVoice.git'
REPO = Path('/content/OneVoice')
DRIVE_ROOT = Path('/content/drive/MyDrive/OneVoice')
if (REPO / '.git').is_dir():
    subprocess.run(['git', '-C', str(REPO), 'pull', '--ff-only', 'origin', 'main'], check=True)
else:
    subprocess.run(['git', 'clone', '--depth', '1', '--branch', 'main', GITHUB_REPO, str(REPO)], check=True)
os.chdir(REPO)
REPORT_ROOT = DRIVE_ROOT / 'reports'
OUTPUT_DIR = REPORT_ROOT / 'onevoice_benchmark_report_v1'
print('Source:', REPO)
print('Input reports:', REPORT_ROOT)
print('Output:', OUTPUT_DIR)
```

```python
command = [sys.executable, 'scripts/build_benchmark_report.py', '--report-root', str(REPORT_ROOT), '--output-dir', str(OUTPUT_DIR)]
print('> ' + ' '.join(command), flush=True)
subprocess.run(command, check=True)
```

```python
from IPython.display import HTML, display
display(HTML((OUTPUT_DIR / 'report.html').read_text(encoding='utf-8')))
print('Markdown:', OUTPUT_DIR / 'report.md')
print('JSON:', OUTPUT_DIR / 'summary.json')
```
